Business Understanding: Identify which segments, categories, and regions drive profit vs loss in the Superstore data, and spot patterns in discounting that hurt margins.

Data Preparation (Phase 3)

Dataset had no missing values or duplicates. Key prep step:
- Converted Order Date and Ship Date from text to datetime, so date-based calculations would be possible if needed later.

(Note: I considered cretaing extra features like Profit Margin and Shipping Duration, but decided the original columns were enough to answer my business question, so I focused analysis on Sales, Profit, Discount, Category, Region, and Segment instead.)

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print(df.dtypes)
```

```python
df.describe()
```

```python
print(df["Category"].unique())
print(df["Sub-Category"].unique())
print(df["Region"].unique())
print(df["Segment"].unique())
```

Cleaning Summary

- No missing values or duplicate rows found
- Order Date and Ship Date converted from text to datetime
- Numeric columns (Sales, Quantity, Discount, Profit) checked 
- No invalid values found
- Categorical columns checked for spelling / consistency - all clean
- Dataset required minimal cleaning; ready for analysis

```python
total_sales = df["Sales"].sum()
print("Total Sales:", total_sales)
```

```python
total_profit = df["Profit"].sum()
print("Total Profit:", total_profit)
```

```python
overall_margin = total_profit / total_sales
print("Overall Profit Margin:", overall_margin)
```

Finding: Overall profit margin is approximately 12.5% - across all sales combined, about 12.5 cents of every $1 in revenue became profit. This is an average acress the whole dataset, not a fixed rate for every product - some categories (like Technology) perform above this average, while others (like Furniture) fall well below it, as explored later in this analysis.

```python
total_orders = df["Order ID"].nunique()
print("Total orders:", total_orders)
```

```python
sales_by_region = df.groupby("Region")["Sales"].sum()
print(sales_by_region)
```

```python
profit_by_region = df.groupby("Region")["Profit"].sum()
print(profit_by_region)
```

Finding: Sales vs Profit by Region

- West: $725,458 sales, $108,418 profit
- East: $678,781 sales, $91,523 profit
- Central: $501,240 sales, $39,706 profit
- South: $391,722 sales, $46,749 profit

WEST is the strongest performer overall - highest sales AND highest profit. 

Interesting anomaly: Central has higher sales than South ($501k vs $392k), but lower profit ($39.7k vs $46.7k). This suggests Central may be over - discounting or selling a less profitable product mix - worth investigating further (checking average discount by region).


```python
discount_by_region = df.groupby("Region")["Discount"].mean()
print(discount_by_region)
```

Confirmed: Central has the highest average discount (24.0%) of all regions, 
nearly double West's (10.9%). This strongly suggests aggressive discounting 
is the reason Central sells more than South but keeps far less profit.

```python
sales_by_category = df.groupby("Category")["Sales"].sum()
print(sales_by_category)
```

```python
profit_by_category = df.groupby("Category")["Profit"].sum()
print(profit_by_category)
```

Finding: Category Performance

- Furniture: $741,999 sales, $18,451 profit (appproximately 2.5% margin)
- Office Supplies: $719,047 sales, $122,491 profit (approximately 17.0% margin)
- Technology: $836,154 sales, $145,455 profit (approximately 17.4% margin)

Furniture has sales comparable to Office Supplies, but profit 6-7x lower. 
Its margin (~2.5%) is far below Office Supplies and Technology (~17% each), 
suggesting heavy discounting or high costs specific to furniture items. 
Worth drilling into Sub-Category next to find exactly which products 
are the problem.

```python
subcategory_profit = df.groupby("Sub-Category")[["Sales", "Profit"]].sum()
print(subcategory_profit)
```

```python
df[df["Sub-Category"] == "Bookcases"]["Category"].unique()
```

```python
subcategory_profit = df.groupby("Sub-Category")[["Sales", "Profit"]].sum()
subcategory_profit_sorted = subcategory_profit.sort_values("Profit")
print(subcategory_profit_sorted)
```

Finding: Loss-making Sub-categories

When i sorted sub-categories by profit, three came out negative:
- Tables: -$17,725 (the biggest loss, even though it had $206,965 in sales)
- Bookcases: -$3,472
- Supplies: -$1,189

Tables and Bookcases are both part of the Furniture category, which explains why Furniture's overall margin (~2.5%) was so much lower than Office Supplies and Technology earlier. Tables especially stands out - it sells a lot, but still loses money on most sales, which suggests it might be discounted too heavily compared to what it actually costs.

```python
segment_summary = df.groupby("Segment")[["Sales", "Profit"]].sum()
print(segment_summary)
```

```python
top_products = df.groupby("Product Name")["Profit"].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
bottom_products = df.groupby("Product Name")["Profit"].sum().sort_values().head(10)
print(bottom_products)
```

Finding: Top and Bottom Products

Looking at individual produts, the  Canon imageCLASS 2200 Advanced Copier was the clear top performer btinging in $25,199 profit — well ahead of every other single product.

On the losing end, the  Cubify CubeX 3D Printer (Double Head) lost the most money at $-8,879, and its Triple Head version  also lost money (-$3,839), which makes me think this entire 3D printer product line is unprofitable.

I also noticed several conference/meeting tables (Chromcraft, Bush Advantage, Balt, BoxOffice) showing up among the worst performers  - which lines up with Tables being the most loss-making sub-category overall.

Summary of Findings

Overall, the Superstore made about $2.3 million in sales, but only $286,397 of that turned into actual profit - so roughly 12.5% of every dollar sold was kept as profit, and therest went to costs.

Breaking it down further, a few things stood out to me:

1. Central region discounts too much. Central gives an average discount of 24%, way higher than every other region (West is only 11%). This is probably why Central has higher sales than South but ends up with less profit - they're giving away too much on each sale.

2. Furniture does'nt make sense financially. Furniture sells almost as much as Office Supplies, but makes 6-7x less profit. When I looked closer, it's mainly two sub-categories - Tables and Bookcases - that are actually losing money, not just making low profit.

3. A couple of specific products are the real problem. The Cubify CubeX 3D Printer (both th double-head and triple-head versions) lost the most money out of every single product. I also noticed several conference tables showing up as losses, which matches the Tables issue above.

4. Everything else (segments, other categories) looked normal - sales and profit moved together the way you'd expect, nothing alarming.
