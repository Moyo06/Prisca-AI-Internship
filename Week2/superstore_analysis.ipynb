Importing the Libraries


```python
import pandas as pd
import matplotlib.pyplot as plt

```

Loading the dataset

```python
df = pd.read_csv("./data/Sample - Superstore.csv", encoding ="latin")
df.head()
```

Checking the size, structure, and quality of the dataset before cleaning.

Data Preparation (Phase 3)

Dataset had no missing values or duplicates. Key prep step:
- Converted Order Date and Ship Date from text to datetime, so date-based calculations would be possible if needed later.

(Note: I considered cretaing extra features like Profit Margin and Shipping Duration, but decided the original columns were enough to answer my business question, so I focused analysis on Sales, Profit, Discount, Category, Region, and Segment instead.)

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print(df.dtypes)
```

```python
df["Year"] = df["Order Date"].dt.year
```

```python
df.describe()
```

Determine the company's overall performance

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("total_sales:", total_sales)
print("total_profit:", total_profit)
```

Compare the performance of each region

```python
region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()
region_analysis
```

```python
region_analysis ["Profit Margin (%)"] = (region_analysis ["Profit"] / region_analysis ["Sales"] * 100 )
region_analysis
```

```python
region_analysis[['Sales', 'Profit']].plot(kind='bar', figsize=(8,5))

plt.title("Sales and Profit by Region")
plt.ylabel("Amount")
plt.show()
```

Key Findings:

- The West region recorded the highest sales and profit.
- The East region closely followed and demonstrated strong overall performance.
- The Central region had the lowest profit margin, indicating that despite generating sales, it converted less revenue into profit.

Recommendation

The company should continue investing in the West while investigating the factors limiting profitability in the Central region.

Sales Trend Analysis

To evaluate how sales changed over time and identify regional growth patterns.

```python
df["Year"] = df["Order Date"].dt.year

year_region = df.groupby(["Year", "Region"])["Sales"].sum().unstack()

year_region
```

```python
growth = ((year_region.loc[2017] - year_region.loc[2014]) / year_region.loc[2014]) * 100

growth
```

```python
year_region.plot(figsize=(10,6), marker='o')

plt.title("Sales trend by Region (2014-2017)")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

Key Findings:

- The East showed consistent year-over-year sales growth from 2014 and 2017.
- The West maintained the highest sales throughout the period despite experiencing a slight decline between 2014 and 2015.
- Central and South also experienced temporary declines before recovering in later years.

Recommendation

Maintain investment in the West while studying the factors driving the East region's consistent growth to replicate that success in other regions.

State Analysis

To identify the states contibuting the most profit and those generating losses.

```python
state_analysis = df.groupby("State")[["Sales", "Profit"]].sum()

top_states = state_analysis.sort_values("Profit", ascending=False).head(10)

top_states
```

```python
bottom_states = state_analysis.sort_values("Profit").head(10)

bottom_states
```

Key Findings:

- California and New York generated the highest profits.
- Texas recorded the largest loss despite having relatively high sales.
- This suggests that high sales do not always translate into high profitability.

Recommendation

Investigate, pricing, discounts, and product mix in loss-making states, particularly Texas, Ohio and Pennsylvania

Customer Segment Analysis

To compare the performance of different customer segments.

```python
customer_segment_analysis = df.groupby("Segment")[["Sales", "Profit"]].sum()

customer_segment_analysis
```

```python
customer_segment_analysis ["Profit Margin (%)"] = (customer_segment_analysis ["Profit"] / customer_segment_analysis ["Sales"]) * 100

customer_segment_analysis
```

Key Findings:

Consumer customers generated the highest sales and total profit , making them the largest contributor to revenue. However, Home Office achieved the highest profit margin, suggesting stronger profitablity relative to its sales.

Recommendation

The company should continue maintaining its focus on the customer segment due to its high sales contribution while developing strategies to expand the Home Office segment because of it higher profitability. The company should also investigate what factors contribute to the stronger margins of Corporate and Home Office segments and apply similar strategies to other segments.


Category Analysis

To compare the performance of the company's major product categories.

```python
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

category_analysis
```

```python
category_analysis.plot(kind='bar', figsize=(8,5))

plt.title("Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")

plt.show()
```

Key Findings:

- Technology had the highest sales and profit.
- Office Supplies had strong profitability despite lower sales.
- Furniture had high sales but the lowest profit.

Recommendation

Continue investing in Technology while reviewing the furniture category to identify ways to improve profitability.

Discount Analysis

To determine how discount levels affect sales and profitability.

```python
discount_analysis = df.groupby('Discount')[['Sales', 'Profit']].sum()

discount_analysis
```

```python
discount_analysis['Profit'].plot(kind='bar', figsize=(10,5))

plt.title('Profit by Discount Level')
plt.xlabel('Discount')
plt.ylabel('Profit')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Discount'], df['Profit'], alpha=0.5)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

Key Findings:

- Lower discounts generated higher profits.
- Discounts above 30% resulted into losses.
- The scatterplot shows a negative relationship between discount and profit, suggesting that increasing discounts can reduce profitability.

Recommendation

The company should avoid excessive discounts and review discount strategies to maintain profitability. 

Shipping Analysis

To determine whether shipping methods and delivery time influence profitability.

```python
shipping_analysis = df.groupby('Ship Mode')[['Sales','Profit']].sum()

shipping_analysis
```

```python
shipping_analysis['Profit Margin (%)'] = (shipping_analysis['Profit'] / shipping_analysis['Sales']) * 100

shipping_analysis
```

```python
shipping_analysis['Profit'].plot(kind='bar', figsize=(8,5))

plt.title("Profit by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Profit")

plt.show()
```

Key Findings:

- Standard Class generated the highest sales and profit due to having large order volumes.
- Profit Margins across shipping modes were relatively similar.
- Shipping mode does not appear to have a major impact on profitability.

Recommendation

The company should continue using current shipping methods while focusing more on pricing and discount strategies to improve profitability.

```python
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days

df[['Order Date','Ship Date','Shipping Days']].head()
```

```python
shipping_time_analysis = df.groupby('Shipping Days')[['Sales','Profit']].sum()

shipping_time_analysis
```

```python
shipping_time_analysis['Profit'].plot(kind='bar', figsize=(10,5))

plt.title("Profit by Shipping Days")
plt.xlabel("Shipping Days")
plt.ylabel("Profit")

plt.show()
```

Key Findings:

- Most sales and profit were generated from orders delivered within 4-5 shipping days.
- Shipping duration does not show a clear negative effect on profitability, as longer delivery times still generated positive profits.
- Other factors such as discounts, product category, and customer- demand may have stronger influence on profit than shipping duration

Recommendation

The company should maintain efficient delivery process but focus more on factors such as discount management and product profitability to improve overall performance.


Correlation Analysis

To ensure the relationships among key numerical variables

```python
correlation = df[['Sales','Quantity','Discount','Profit']].corr()

correlation
```

```python
import matplotlib.pyplot as plt

corr = df[['Sales','Quantity','Discount','Profit']].corr()

plt.figure(figsize=(6,5))
plt.imshow(corr, cmap='coolwarm')

plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```

Key Findings:

- Sales and Profit have a moderate positive relationship, meaning higher sales generally tend to result in higher profit.
- Discount has a negative relationship with Profit, suggesting that higher discounts may reduce profitability.
- Quantity has a very weak relationship with Profit, indicating that selling more items does not always lead to higher profit.

Recommendation

The company should focus on improving profitable sales and carefully manage discount strategies to avoid reducing profit margins.

Conclusion

This analysis examined the factors affecting sales and profitability in the Sample Superstore dataset. The findings showed that the West region had the highest overall performance, while the East region showed strong and consistent growth. Technology was the strongest-performing category, while Furniture generated high sales but lower profitability compared to other categories.

The analysis also showed that high discount levels negatively affected profit, while shipping factors had little effect on profitability. Overall, focusing on profitable sales, improving discount strategies, and identifying growth opportunities can help improve business performance.

Overall Recommendations

- Carefully manage discount levels to avoid reducing profitability.
- Continue investing in high-performing categories such as Technology.
- Focus on profitable growth by studying successful regions like West and East.
- Consider profitability, not only sales volume, when making business decisions.
